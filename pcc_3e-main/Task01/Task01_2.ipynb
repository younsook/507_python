```python
class Adder:
    def __init__(self, a, b):
        self.a =a
        self.b = b
    def sum(self):
        return self.a + self.b
    def __call__(self):
        return self.a + self.b

add = Adder(1,2)
add.sum()
```

```python
add = Adder(1,2)
add()
```

```python
class MySum:
    def __init__(self, a: list):
        self.data = a

    def __call__(self, *args, **kwds):
        return sum(self.data)

    def __getitem__(self, idx: int):
        return self.data[idx]

s = MySum([1,2,3,4])
s[0] + s[1]
```

```python
s = MySum([1,2,3,4])
s[0] + s[1]
```

```python
r = range(3)          # [0, 1, 2] 반복 가능한 객체
it = iter(r)          # 이터레이터로 변환

print(next(it))  # 0
print(next(it))  # 1
print(next(it))  # 2
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}({self.age}.)"

p = Person("윤", 42)
print(p)            # 자동으로 __str__ 호출됨
print(str(p))       # 명시적으로 __str__ 호출
```
