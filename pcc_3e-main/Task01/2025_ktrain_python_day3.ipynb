## 실습 문제 1: 일차 함수

함수 f(x) = 2x + 1 을 구현하고 그래프를 그려보세요.

- **요구사항:**

    0부터 5까지의 x 값에 대해 y 값을 계산하는 f(x) 함수를 순수 Python으로 작성하세요. (힌트: range 와 list 사용)



```python


def f(x):
    return 2 * x + 1

xValues = list(range(6))

# 각 x에 대한 y값 계산
yValues = [f(x) for x in xValues]

print("x: ", xValues)
print("y: ", yValues)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return 2 * x + 1

xValues = list(range(6))

# 각 x에 대한 y값 계산
yValues = [f(x) for x in xValues]

plt.plot(xValues, yValues, marker='D', linestyle='--', color='green')
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.title("Graph of f(x)= 2x + 1 일차 함수")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

## 실습 문제 2: 이차 함수 (다항 함수)

함수 f(x) = 2x^2 - 3x + 4 를 구현하고 그래프를 그려보세요.

- **요구사항:**

    5부터 5까지의 x 값에 대해 y 값을 계산하는 f(x) 함수를 순수 Python으로 작성하세요.

```python
def f(x):
    return 2* x**2 - 3*x + 4

xValues = list(range(6))

yValues = [f(x) for x in xValues]

print("xValues: ", xValues)
print("yValues: ", yValues)
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 마이너스 깨짐 방지

def f(x):
    return 2* x**2 - 3*x + 4

xValues = list(range(-10, 11))

yValues = [f(x) for x in xValues]

plt.plot(xValues, yValues)
plt.title("f(x) = 2x^2 - 3x + 4 이차 함수 (다항 함수)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show() #.py에서는 show없으면 출력안됨. 
```
