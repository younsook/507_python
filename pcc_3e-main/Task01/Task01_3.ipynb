filter(func, iterable)	조건을 만족하는 요소만 추출

```python
nums = [1, 2, 3, 4, 5, 6]

def is_even(x):
    return x % 2 == 0 # 짝수만 필터링

result = filter(is_even, nums)
print(list(result)) 
```

```python
nums = [1, 2, 3, 4, 5, 6]

# 짝수만 필터링
evens = list(filter(lambda x: x % 2 == 0, nums))

print(evens)
```

```python
from functools import reduce
from operator import add

result = reduce(add, nums, 0)
print(result)  # 출력: 21
```

```python
# id()를 사용하여 객체의 주소(참조값) 를 출력하고 비교하는 예제
a = [1, 2, 3]
b = a        # 같은 객체
c = [1, 2, 3] # 새로운 객체 (내용만 같음)

print(f"a: {a}, id(a): {id(a)}")
print(f"b: {b}, id(b): {id(b)}")
print(f"c: {c}, id(c): {id(c)}")

print("a is b:", a is b)  # True (같은 객체)
print("a is c:", a is c)  # False (다른 객체)
print("a == c:", a == c)  # True (값은 같음)
```

```python
# isinstance()는 파이썬에서 값이 특정 자료형(또는 클래스의 인스턴스) 인지를 확인할 때 사용하는 함수
a = 10
b = "hello"
c = [1, 2, 3]

print(isinstance(a, int))      # True
print(isinstance(b, str))      # True
print(isinstance(c, list))     # True
print(isinstance(c, dict))     # False
```

```python
class CountUpTo:
    def __init__(self, max):
        self.max = max
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.max:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1
        

counter = CountUpTo(3)

for num in counter:
    print(num)
```
