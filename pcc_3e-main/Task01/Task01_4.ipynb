```python
class Number:
    
    def __init__(self, value):
        self.value = value  
        
    def __add__(self, other):
        return Number(self.value + other.value)
    
    def __str__(self):
        return str(self.value)
    
    def __repr__(self):
        return self.__str__()
```

```python
a = Number(10)
b = Number(5)
c = a + b

print(c) 
```

```python
class EnenNumber(Number):
    def __init__(self, value):
        super().__init__(value)
        if self.value % 2  != 0  :
            self.value = value + 1
```

```python
n1 = EnenNumber(5)
n2 = EnenNumber(10)

n1 + n2
```

```python
from abc import ABC, abstractmethod

class Player(ABC):
    def __init__(self, name: str):
        self.name = name  # 인스턴스 변수 초기화

    @abstractmethod
    def run(self):
        pass
```

```python
class SoccerPlayer(Player):
    def run(self):
        print(f"⚽ {self.name}, 축구장을 전력 질주합니다!")

class BasketballPlayer(Player):
    def run(self):
        print(f"🏀 {self.name}, 코트를 빠르게 달립니다!")
```

```python
p1 = SoccerPlayer("손흥민")
p2 = BasketballPlayer("이현중")

p1.run()  # ⚽ 손흥민 축구장을 전력 질주합니다!
p2.run()  # 🏀 이현중 코트를 빠르게 달립니다!
```
