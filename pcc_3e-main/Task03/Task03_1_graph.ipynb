```python
import numpy as np
import matplotlib.pyplot as plt

# 함수 정의: f(x) = x^2 + 2x + 1
def f(x):
    return x**2 + 2*x + 1

# x의 범위: -5부터 5까지
x = np.arange(-5, 6)  # [-5, -4, ..., 5]
x = np.linspace(-6, 6,100)
y = f(x)

# 그래프 그리기
plt.plot(x, y, marker='o', label='f(x) = x² + 2x + 1')
plt.axhline(0, color='gray', linestyle='--', linewidth=0.5)  # x축
plt.axvline(0, color='gray', linestyle='--', linewidth=0.5)  # y축
plt.title("Graph of f(x) = (x + 1)^2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()


```

```python
from random import choice


class RandomWalk:
    """A class to generate random walks."""

    def __init__(self, num_points=5000):
        """Initialize attributes of a walk."""
        self.num_points = num_points

        # All walks start at (0, 0).
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """Calculate all the points in the walk."""
        # Keep taking steps until the walk reaches the desired length.
        while len(self.x_values) < self.num_points:

            # Decide which direction to go, and how far to go.
            x_direction = choice([1, -1])
            x_distance = choice([0, 1, 2, 3, 4])
            x_step = x_direction * x_distance

            y_direction = choice([1, -1])
            y_distance = choice([0, 1, 2, 3, 4])
            y_step = y_direction * y_distance

            # Reject moves that go nowhere.
            if x_step == 0 and y_step == 0:
                continue

            # Calculate the new position.
            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)
```

```python
rw = RandomWalk()
rw.fill_walk()
```

```python
#plt.scatter(rw.x_values, rw.y_values)
import matplotlib.pyplot as plt
from random import choice

# 랜덤 워크 클래스 정의
class RandomWalk:
    def __init__(self, num_points=5000):
        self.num_points = num_points
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        while len(self.x_values) < self.num_points:
            x_direction = choice([1, -1])
            x_distance = choice([0, 1, 2, 3, 4])
            x_step = x_direction * x_distance

            y_direction = choice([1, -1])
            y_distance = choice([0, 1, 2, 3, 4])
            y_step = y_direction * y_distance

            if x_step == 0 and y_step == 0:
                continue

            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)

# 인스턴스 생성 및 데이터 생성
rw = RandomWalk()
rw.fill_walk()

# 산점도 시각화
plt.figure(figsize=(10, 6))



# 전체 경로 표시 (점마다 색을 다르게)
point_numbers = list(range(rw.num_points))
plt.scatter(rw.x_values, rw.y_values, c=point_numbers, cmap=plt.cm.Blues, edgecolors='none', s=10)

# 시작점은 초록색
plt.scatter(0, 0, c='green', edgecolors='none', s=50)

# 마지막 점은 빨간색
plt.scatter(rw.x_values[-1], rw.y_values[-1], c='red', edgecolors='none', s=50)

# 축 제거
plt.axis('off')
plt.title("Random Walk (시작: 초록 / 종료: 빨강)")
plt.show()
```

```python
from random import randint

class Die:
    """A class representing a single die."""

    def __init__(self, num_sides=6):
        """Assume a six-sided die."""
        self.num_sides = num_sides

    def roll(self):
        """"Return a random value between 1 and number of sides."""
        return randint(1, self.num_sides)
    
    def roll_many(self, times):
        return [self.roll() for _ in range(times)]
```

```python
d1 = Die()
d1.roll_many(100)
```

```python
plt.hist(d1.roll_many(1000))
```

```python
d1 = Die()
d1.roll_many(1000)
d2 = Die()
d2.roll_many(1000)
```

```python
# 먼저 각각 주사위를 여러 번 굴린 결과 리스트를 저장
results1 = d1.roll_many(1000)
results2 = d2.roll_many(1000)

# 이제 zip을 사용해 같은 인덱스의 결과를 합산
data = [a + b for a, b in zip(results1, results2)]
print(data[:10])
```
